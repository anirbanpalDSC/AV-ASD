```python
import pandas as pd
import shutil
import os

# Path settings
csv_path = "C:\\Users\\anirb\\Downloads\\annotated_with_prompt.csv"
source_folder = "C:\\Users\\anirb\\Downloads\\asd_demo_data"
target_folder = "C:\\Users\\anirb\\Downloads\\ap"    # Folder to copy to

# Create destination folder if not exists
os.makedirs(target_folder, exist_ok=True)

# Read CSV
df = pd.read_csv(csv_path, header=None)   # header=None if no column name
file_list = df[1].tolist()

for file_name in file_list:
    # Build full file paths
    src = os.path.join(source_folder, file_name)
    dst = os.path.join(target_folder, file_name)

    # Check if file exists then copy
    if os.path.isfile(src):
        shutil.copy(src, dst)
        print(f"✅ Copied: {file_name}")
    else:
        print(f"❌ Not Found: {file_name}")
```
